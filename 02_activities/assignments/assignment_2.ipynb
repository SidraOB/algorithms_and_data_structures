```python
# Your answer here
# Definition for a binary tree node.
class TreeNode(object):
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

def print_preorder(root):
  if root:
    print(root.val, end = " ")
    print_preorder(root.left)
    print_preorder(root.right)

def bt_path(root: TreeNode):
  all_paths = []
  bt_path_helper(root, all_paths, [])
  return all_paths

def bt_path_helper(root, lst, current_path):
    current_path = current_path.copy()
    if root:
      current_path.append(root.val)
      if not root.left and not root.right:
        lst.append(current_path)
      else:
        bt_path_helper(root.left, lst, current_path)
        bt_path_helper(root.right, lst, current_path)
```


-   Explain why their solution works in your own words.


This solution works by recursively checking all possible paths from the root to the leaves in the binary tree. It detects leaf nodes (nodes with no children) and appends the corresponding path to the result only when it reaches a leaf. The use of curent_path.copy ensures that paths are tracked correctly.






-   Explain the problem’s time and space complexity in your own words.


### Time complexity
In the worst case, where tree is not balanced, the time complexity can become O(n * n) = O(n²).However, in the best case (a balanced tree), time complexity will be O(n log n).

### Space Complexity:
In the worst case, where tree is not balanced, this becomes O(n * n) = O(n²).In the best case (balanced tree), this is O(n log n).
